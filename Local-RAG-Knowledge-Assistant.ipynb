```python
import ollama
```

```python
#load Dataset
with open('text.txt') as file:
    dataset=file.readlines()
print(dataset)
print(len(dataset))
```

```python
dataset
```

```python
EMBEDDING_MODEL = 'hf.co/CompendiumLabs/bge-base-en-v1.5-gguf'
LANGUAGE_MODEL = 'hf.co/bartowski/Llama-3.2-1B-Instruct-GGUF'
```

```python
vectorDB=[]
def add_chunk_db(chunk):
    embedding=ollama.embed(model=EMBEDDING_MODEL,input=chunk)['embeddings'][0]
    print(embedding)
    vectorDB.append((chunk,embedding))
```

```python
for chunk in dataset:
    add_chunk_db(chunk)
```

```python
for i in vectorDB:
    print(i)
```

```python
def cosine_similarity(a,b):
    dot_product=sum([x*y for x,y in zip(a,b)])
    norm_a=sum([x**2 for x in a])**0.5
    norm_b=sum([y**2 for y in b])**0.5
    return dot_product/(norm_a*norm_b)
```

```python
def retrieve(query,top=3):
    query_embedding=ollama.embed(model=EMBEDDING_MODEL,input=query)['embeddings'][0]
    #print(query_embedding)
    similarities=[]
    for chunk,embedding in vectorDB:
        similarity=cosine_similarity(query_embedding,embedding)
        #print(chunk,similarity)
        similarities.append((chunk,similarity))

    similarities.sort(key=lambda x:x[1],reverse=True)
    #print(similarities)
    return similarities[:top]
```

```python
input_query=input("USER:")
retrieve_know=retrieve(input_query,2)
retrieve_know
```

```python
instruction_prompt=f"""you are a helpful chatbot.
Use only the given context of texts to answer the question. 
do not create new information,{' '.join(know[0] for know in retrieve_know)}"""
```

```python
instruction_prompt
```

```python
stream=ollama.chat(
    model=LANGUAGE_MODEL,
    messages=[
        {'role':'system','content':instruction_prompt},
        {'role':'user','content':input_query}
    ],
    stream=True
)
```

```python
for chunk in stream:
    print(chunk['message']['content'],end='')
```
